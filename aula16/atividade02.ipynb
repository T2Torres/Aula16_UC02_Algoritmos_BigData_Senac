```python
import pandas as pd
import numpy as np
import traceback
import matplotlib.pyplot as plt
```

```python
ENDERECO_DADOS = 'https://www.ispdados.rj.gov.br/Arquivos/BaseDPEvolucaoMensalCisp.csv'

# Carregando dados
df_ocorrencias = pd.read_csv(ENDERECO_DADOS, sep=';', encoding='iso-8859-1')
    
    
# Delimitando as variáveis 'cisp', 'roubo_veiculo', 'recuperacao_veiculos'
df_ocorrencias = df_ocorrencias[['cisp', 'roubo_veiculo', 'recuperacao_veiculos']]

# Agrupar as ocorrências por 'cisp' e somar os valores de 'roubo_veiculo' e 'recuperacao_veiculos'
df_total_veiculos = df_ocorrencias.groupby('cisp').agg({'roubo_veiculo': 'sum', 'recuperacao_veiculos': 'sum'}).reset_index()
    
    
    
    
print(df_total_veiculos.head(10))  
```

```python
print('Analisando dados...')
# Excluir uma parte de valores extremos
df_total_veiculos_cut = (df_total_veiculos[df_total_veiculos['roubo_veiculo'] < np.percentile(df_total_veiculos['roubo_veiculo'], 95)])


df_total_veiculos_cut = (df_total_veiculos[df_total_veiculos['recuperacao_veiculos'] < np.percentile(df_total_veiculos['recuperacao_veiculos'], 99)])

# Transformando em arrays
array_roubo_veiculo = np.array(df_total_veiculos_cut['roubo_veiculo'])
array_recuperacao_veiculos = np.array(df_total_veiculos_cut['recuperacao_veiculos'])

# Calculando a correlação entre roubo de veículos e recuperação de veículos
correlacao = np.corrcoef(array_roubo_veiculo, array_recuperacao_veiculos)[0, 1]
print(f'Correlação entre roubo de veículos e recuperação de veículos: {correlacao:.2f}')
```
